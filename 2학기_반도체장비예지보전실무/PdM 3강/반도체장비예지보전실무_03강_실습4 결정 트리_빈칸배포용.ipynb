## 데이터 로딩 및 전처리

### 데이터 로딩

```python
import pandas as pd

# 데이터 로딩
df = pd.read_csv('chemical_data_lec03.csv')
```

### 범주형 데이터 처리

```python
# 라벨(Failure) 데이터의 현재 상태 확인
df.Failure.value_counts()
```

```python
# Failure 값 맵핑
df.Failure = df.Failure.map({'정상': 0, '0': 0,
                '불량': 1, '1': 1})

# 결과 확인
df.info()
df.Failure.value_counts()
```

### 결측치 처리

```python
# 결측치 확인
df.isnull().sum()
```

```python
# 결측치 처리
# 중앙값으로 대체하자.
temp_median = df.Temp.median()
df.Temp = df.Temp.fillna(temp_median)

visco_median = df.Viscosity.median()
df.Viscosity = df.Viscosity.fillna(visco_median)
```

```python
df.info()
```

### 중복 데이터 처리

```python
# 중복값이 있는지 확인
df.Lot.duplicated().sum()
```

```python
# 중복값 삭제
df = df.drop_duplicates(subset=['Lot'], keep='first')
```
