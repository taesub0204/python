```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense

# 1. Fashion MNIST 데이터 불러오기
(X_train, y_train), (X_test, y_test) = \
    tf.keras.datasets.fashion_mnist.load_data()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

# 2. 이미지 데이터 정규화
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# 3. 다층 신경망 모델 구성
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation='relu'),   # 은닉층
    Dense(10, activation='softmax')  # 출력층
])

model.summary()
```

```python
# 4. 모델 학습 방법 설정
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 5. 모델 학습
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
# 6. 테스트 데이터로 모델 평가
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("테스트 정확도:", test_accuracy)
```

```python
import numpy as np

# 7. 테스트 데이터 예측
predictions = model.predict(X_test)

predicted_class = np.argmax(predictions[0])

print("예측 클래스:", predicted_class)
print("실제 클래스:", y_test[0])
```

## Optional

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]

print("예측:", class_names[predicted_class])
print("정답:", class_names[y_test[0]])
```
