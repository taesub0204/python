# 파이썬 클래스

- 클래스는 붕어빵 틀, 자동차 설계도 등
- 객체는 붕어빵, 실제 자동차(설계도로부터 찍어낸 실체)
- 인스턴트는 설계도에서 실제 자동차를 만드는 것

````python
# 클래스 예제 1

class fishbread:
    def __init__(self, filling):
        self.filling = filling
    def bake(self):
        print(f"{self.filling} 붕어빵이 구워졌습니다.")

bread1 = fishbread('팥')
bread2 = fishbread('슈크림')
bread1.bake()
bread2.bake()
````

````python
# 클래스 예제 2

class Myclass:
    def set_info(self, name, age):
        self.name = name
        self.age = age
    def print_info(self):
        print(f'내 이름은 {self.name}입니다.')
        print(f'내 나이는 {self.age}입니다.')

a = Myclass()
a.set_info('홍길동', 25)
a.print_info()
````

````python
# 클래스 예제 3

class Myclass:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def print_info(self):
        print(f'내 이름은 {self.name}입니다.')
        print(f'내 나이는 {self.age}입니다.')

a = Myclass('홍길동', 25)
a.print_info()
````

## self 개념

- self는 인스턴스 자신을 가리키는 키워드
- __init__()는 생성자 메서드(객체 생성시 자동 실행)


### 예제 2와 예제 3의 차이

```python
# __init__ 없을 때
a = Myclass()               # 빈 객체 생성
a.set_info("홍길동", 25)    # set_info 따로 초기화
a.print_info()
```
- Myclass() 호출 시 아무일도 일어나지 않음
- 따로 set_info() 메서드를 명시적으로 호출해야 정보 초기화 일어남
- 실수로 set_info() 호출을 하지 않으면 print_info()에서 에러 발생
- 객체 사용 전, 무조선 초기화 메서드를 호출해야 하는 단점이 있음

```python
# __init__ 있을 때
a = Myclass("홍길동", 25)   # 생성 시점에 바로 초기화
a.print_info()
```
- 객체 생성(Myclass(...))과 동시에 __init__() 실행
- 객체가 "초기화괸 상태"로 바로 준비되므로 안전하고 직관적
- 실수를 줄이고, 코드 가독성과 재사용성이 좋아짐
- 객체를 만들때, 생성자에서 바로 초기화 &rarr; 객체의 상태 일관성 보장

````python
# 클래스 예제 2 실수 버전

class Myclass:
    def set_info(self, name, age):
        self.name = name
        self.age = age
    def print_info(self):
        print(f'내 이름은 {self.name}입니다.')
        print(f'내 나이는 {self.age}입니다.')

a = Myclass()
a.set_info('홍길동', 25)
a.print_info()   # AttributeError! set_info() 안 했기 때문
````
