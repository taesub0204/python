## IF - ELSE 조건문
- 조건문이 참이라면, if문 실행
- 거짓이라면, else문 실행

```python
if 조건문:
    조건이 참일 때 실행
else:
    조건이 거짓일 때 실행
```

````python
a = False        # True : 참

if a:
    print("a는 True입니다.")
    print("if문에 속해있는 문장입니다.")
else:
    print("a는 False입니다.")
    print("else문에 속해있는 문장입니다.")
````

````python
# 점수가 60점 이상이면, 합격을 통보
# 점수가 60점 미만이면, 불합격을 통보

score = 90

if score >= 60:
    print("합격입니다.")
else:
    print("불합격입니다.")
````

## in 연산자
문자열, 리스트, 튜플에 원하는 원소가 있는지 확인하는 연산자

````python
list1 = [1, 2, 3, 4, 5]
tuple1 = (2, 3, 4, 5)

print(1 in list1)
print(1 in tuple1)
````

````python
"""
주머니에 돈이 있으면 택시를 타자
주머니에 돈이 없으면 걸어서 가자
"""

pocket = ['phone', 'wallet', 'money', 'IDcard']

if 'money' in pocket:
    print("택시를 타자")
else:
    print("걸어가자")

# not in을 활용한 방법
if 'money' not in pocket:
    print("걸어가자")
else:
    print("택시를 타자")
````

## IF - ELIF - ELSE 조건문

```python
if 조건문1:
   조건문1이 참일때 실행되는 문장
...
elif 조건문2:
    조건문1이 거짓이고, 조건문2는 참일 때 실행되는 문장.
elif 조건문3:
    조건문1,2가 거짓이고 조건문 3은 참일 때 실행되는 문장.
else:
    조건문이 모두 거짓일 때 실행되는 문장
```

````python
"""
주머니에 돈과 카드가 모두 존재하면 택시를 타자
주머니에 카드만 있다면 택시를 타자
주머니에 돈만 있다면 버스를 타자
주머니에 돈도 없고 카드도 없다면 걸어가자
"""
pocket = ['phone','IDcard', 'money1', 'card']

# if else 문은 if, elif, else문중 단 하나라도 참이되어 실행되면, 바로 종료한다.

if 'money' in pocket and 'card' in pocket:
  print('둘다 있으니 택시를 타자')
elif 'money' in pocket:
  print('돈만 있으니 버스를 타자')
elif 'card' in pocket:
  print('카드만 있으니 택시를 타자')
else:
  print('아무것도 없으니 걸어가자.')
````

## 중첩 조건문
- 조건문 안에 조건이 들어갈 수 있음

- 중첩 조건문은 조건의 계층적 분기를 표현할 때 사용하지만 너무 많아지면 가독성이 떨어지므로 이후 논리 연산자로 대체 가능

````python
score = 92

if score >= 60:
    print("합격")
    if score >= 90:
        print("우수")
else:
    print("불합격")
````
