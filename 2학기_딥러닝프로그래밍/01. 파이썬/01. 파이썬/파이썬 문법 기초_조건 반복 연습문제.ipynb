# 연습 문제 1

학생들에게 점수에 따라서 등급을 부여하려고 한다.

90점 이상일 경우 A등급.

80점 이상일 경우 B등급.

70점 이상일 경우 C등급.

70점 미만일 경우에는 D등급을 부여하는 코드를 작성하시오.

```python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("D")
```

```python
score = int(input("점수를 입력하세요: "))

if score >= 90:
  print("A등급입니다.")
elif score >= 80:
  print("B등급입니다.")
elif score >= 70:
  print("C등급입니다.")
else:
  print("D등급입니다.")

############좋지 않은 예시#################
if score >= 70:
  print("C등급입니다.")
elif score >= 80:
  print("B등급입니다.")
elif score >= 90:
  print("A등급입니다.")
else:
  print("D등급입니다.")
```

```python
a = int(input("정수를 입력하세요: "))
b = int(input("정수를 입력하세요: "))

if a > b:
    print(a)
elif a < b:
    print(b)
else:
    print("같습니다.")
```

# 연습 문제 3

커피 자판기 예제

1. 커피는 3개가 있다.
2. 커피 하나의 가격은 300이다.

```python
coffee = 3

while True:
    money = int(input("돈을 넣어주세요: "))

    if money == 300:
        print("커피를 줍니다.")
        coffee = coffee - 1
        print("남은 커피의 양은 %d개 입니다." % coffee)
        print('-' * 20)
    elif money > 300:
        print("거스름돈 %d를 주고 커피 제공." % (money - 300))
        coffee = coffee - 1
        print("남은 커피의 양은 %d개 입니다." % coffee)
        print('-' * 20)
    else:
        print("돈을 다시 돌려주고 커피를 주지 않습니다.")
        print("남은 커피의 양은 %d개 입니다." % coffee)
        print('-' * 20)

    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        print('-' * 20)
        break

```

# 연습 문제 4

별찍기 문제

정수를 입력받아 정수크기의 마름모를 찍어보세요.




```python
"""
ex.
입력 정수: 5
출력결과:
   *
  ***
 *****
  ***
   *

"""

# for문으로 짜기
diamond = int(input('별 최대 개수 : '))

if diamond % 2 == 1:
    i = 1
else :
    i = 2

for i in range(i, diamond+1, 2):
    blank = ' '*((diamond-i)//2)
    star = '*'*i # 여기에 * 대신 다양한 기호 사용 가능!

    print(blank, star,blank)

for i in range(i-2, 0, -2):
    blank = ' '*((diamond-i)//2)
    star = '*'*i

    print(blank, star,blank)
```

```python
"""
ex.
입력 정수: 5
출력결과:
   *
  ***
 *****
  ***
   *

"""

# while문 사용
diamond = int(input('별 최대 개수 : '))

if diamond % 2 == 1:
    i = 1
else :
    i = 2


while i < diamond:
    blank = ' '*((diamond-i)//2)
    star = '*'*i
    print(blank, star)

    i += 2

while i > 0:
    blank = ' '*((diamond-i)//2)
    star = '*'*i
    print(blank, star)

    i -= 2
```

```python
n = int(input("홀수 정수를 입력해주세요: "))

if n % 2 == 0:
    print("홀수만 입력이 가능합니다.")
else:
    mid = n // 2
    for i in range(n):
        if i <= mid:
            stars = 2 * i + 1
        else:
            stars = 2 * (n - i - 1) + 1
        spaces = (n - stars) // 2
        print(" " * spaces + "*" * stars)
```
