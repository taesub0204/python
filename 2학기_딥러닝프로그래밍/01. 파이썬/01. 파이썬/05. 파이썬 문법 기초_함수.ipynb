# 함수(function)

> 함수는 자주 사용하는 코드를 '묶어서 저장'하고, 필요할 때 불러서 사용하는 이름표

- 반복되는 코드를 줄임
- 코드를 구조화하여 가독성 향상
- 재사용이 가능

## 1. 기본 구조
```python
def 함수이름(매개변수1, 매개변수2, ...):
    실행할 코드
    return 결과값
```

- def: define(정의)의 줄임말
- 함수이름: 호출할 때 사용할 이름
- 매개변수(parameter): 함수가 받아서 처리할 값(선택 사항)
- return: 함수를 실행한 결과를 돌려주는 키워드(선택 사항)

````python
def say_hello():            # 함수 선언
    print("Hello world")    # 호출하면 실행한 코드

say_hello()                 # 함수 호출
````

- 함수는 만들기만 하면 실행되지 않음
- say_hello()를 호출 해야 실제로 작동

````python
def add(a, b):              # a, b를 외부에서 값을 가져와 사용함(매개변수)
    return a + b            # return: 반환

num1 = 40
num2 = 50

result = add(num1, num2)    # num1(인수)
                            # num2(인수)
print(result)
30

# 여러 값을 반환할 때는 튜플형식으로 반환된다.
def test(a, b):
    return a, b

print(test(1, 2))
# (1, 2)
````

## 지역변수와 전역변수

- 지역변수(local variable)
    - 함수 내부에서 정의된 변수
    - 함수 안에서만 유효
- 전역변수(global variable)
    - 함수 외부에서 정의된 변수
    - 코드 전체에서 유효



````python
# 지역변수 예시
def greet():
    message = "Hello, world!"
    print(message)

greet()
# print(message)
# NameError -> message가 함수 안에서 정의되었기 때문에 함수 밖에서는 알 수 없음
````

````python
# 전역변수 예시
name = "world"

def greet():
    print("Hello", name)

greet()

````

````python
# 전역변수를 함수 안에서 수정하려 할때

counter = 0

def increment():
    counter += 1
    print(counter)

increment()

# UnboundLocalError
# 함수 안에서 수정(counter += 1)을 하려는 순간 파이썬은 counter를 지역변수로 간주하려 시도
# 지역에 선언된 counter가 없기 때문에 오류 발생
````

````python
# 오류 해결 방법
counter = 0

def increment():
    global counter # 전역변수를 사용하겠다고 명시
    counter += 1
    print(counter)

increment()
````

## 지역, 전역변수 요약

| 상황         | 변수 종류 | 접근     | 수정              |
| ---------- | ----- | ---------- | ------------------- |
| 함수 외부에서 선언 | 전역변수  | 어디서든 읽기 가능 | 함수 안에서는 `global` 필요 |
| 함수 내부에서 선언 | 지역변수  | 그 함수 안에서만 사용  | 외부에서 접근 불가          |



## 정리

함수는 입력값을 가지고, '기능'에 맞는 출력값을 내어놓는 것

- 함수는 매개변수와 return이 있을 수 도, 없을 수 도 있음

#### 함수를 사용하는 이유
1. 코드 재사용
2. 코드 정리와 관리가 쉬움
3. 기능 단위로 나누면 협업과 유지보수 유리
    - 하나의 프로그램을 작은 조각(함수)들로 나누어 관리할 수 있음
        - 예: 로그인 함수, 계산 함수, 출력 함수

````python
# 1. 코드 재사용
# 같은 코드 3번 복붙 대신
print("=" * 20)
print(" 학생 관리 프로그램 ")
print("=" * 20)


def print_line():
    print("=" * 20)

print_line()
print(" 학생 관리 프로그램 ")
print_line()
````

````python
# 2. 코드 정리와 관리가 쉬움
# 함수가 없는 경우
a = int(input())
b = int(input())

print("결과: ", a + b)
````

````python
# 2. 코드 정리와 관리가 쉬움
# 함수로 정리
def get_sum():
    a = int(input())
    b = int(input())
    print("결과: ", a + b)

get_sum()
````

 - 곱셈 : mul(a, b)

 두개의 수는 사용자에게 입력받으세요.

```python
a = int(input("첫 번째 숫자를 입력하세요 : "))

```

````python
# 연습 문제 1

a = int(input("첫 번째 숫자를 입력하세요 : "))
b = int(input("두 번째 숫자를 입력하세요 : "))

def add(a, b):
  result = a+b
  print("%d + %d = %d" % (a, b, result))
  return result

def diff(a, b):
  result = a-b
  print("%d - %d = %d" % (a, b, result))
  return result

def div(a, b):
  result = a/b
  print("%d / %d = %d" % (a, b, result))
  return result

def mul(a, b):
  result = a*b
  print("%d X %d = %d" % (a, b, result))
  return result

add(a, b)
diff(a, b)
div(a, b)
mul(a, b)
````

````python
# 연습 문제 2
# 3가지의 수를 입력받아 최대, 최소값을 출력 및 반환하는 함수

def min_max(a, b, c):
  max = a
  min = a

  if b > a:       # b > a, min = a, max = b
    max = b
  else:           # a <= b, min = b, max = a
    min = b

  if c > max:
    max = c

  if c < min:
    min = c

  print("max : %d\nmin : %d" % (max, min))
  return max, min

max_num, min_num = min_max(3, 1, 2)
````
