# 파이썬 변수

## 변수란?
> "변수는 파이썬의 모든 흐름의 출발"

## 변수값을 저장하는 메모리 공간
- 변수 이름과 값을 할당하여 생성
- 이름을 붙여야 그 값을 나중에 찾아 사용할 수 있음

```python
age = 13

print(age)
```

&rarr; `age`는 변수명

&rarr; `13`은 할당된 값

&rarr; `age`라는  이름으로 `13`이라는 값을 사용할 수 있게 됨

> 변수는 "컴퓨터 안에 데이터를 저장해두고 꺼내 쓰기 위한 이름표"

---
## 왜 변수가 필요할까?

```python
print(20)
print(20)
print(20)
```

&rarr; 매번 똑같은 숫자를 사용, 나중에 바꾸고 싶으면 3줄 전부 변경해야함.
하지만

```python
age = 20

print(age)
print(age)
print(age)
```

&rarr; `age` 값만 바꾸면 한꺼번에 변경 가능

> 변수는 효율적인 코드 작성과 유지보수에 필수

---

### 변수에 할당된 값은 프로그램이 실행되는 동안 몇번이나 값이 바뀔 수 있다.

```python
a = 10

print(a)

b = 20

print(b)
```

```python
print(a)

a = 20

print(a)

b = 10

print(b)
```

```python
name = "Alice"

score = 95

is_passed = True
```

|요소|설명

|name, scroe, is_passed|변수 이름|

|"Alice", 95, Ture|실제 값(데이터)|

## 변수 이름 규칙(매우 중요)

#### 사용 가능한 문자:
- 영어 소문자/대문자(`a~z`, `A~Z`)
- 숫자(`0~9`) &rarr; 단, _시작할 수 없음_
- 밑줄 `_`

#### 안되는 예시:
- `1name` &rarr; 숫자로 시작X
- `def`, `class`, `if` 등 &rarr; 예약어 X
- `user-name` &rarr; `-`는 허용하지 않음
