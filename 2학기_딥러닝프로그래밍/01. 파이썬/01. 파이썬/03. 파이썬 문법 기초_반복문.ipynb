# 반복문

> "반복 가능한 대상(iterable)"을 순차적으로 꺼내며 자동화된 흐름을 구성하는 제어 도구

- For, While의 구문 구조 이해
- 반복 가능 대상의 개념
- 반복 흐름 추적
- 들여쓰기의 중요성
    - 파이썬의 블록 구조를 지배하는 들여쓰기 규칙 숙지
    - 파이썬은 {} 중괄호를 사용하지 않음
    - 들여쓰기(공백 4칸)가 블록을 정의하는 유일한 방법
---
- 반복문은 비슷한 코드를 여러번 반복 실행하게 해주는 구조
- "for i in ..." or "while ..."을 사용하여 흐름을 제어

## for문의 기본 구조
```python
for 변수 in 반복가능한 대상:
    수행할 코드 1
    수행할 코드 2
```

````python
for i in [1, 2, 3]:
    print(i)

# 1
# 2
# 3
````

|list, tuple|O|[1, 2, 3], (1, 2, 3)|
|Str|O|"abc" &rarr; 'a', 'b', 'c'|
|range|O|range(5)|
|int, float|X|반복 불가능|

&rarr; 반복문은 반드시 `iterable`한 대상이 필요

````python
list1 = ['one', 'two', 'three', 'four', 'five']

count = 1
for i in list1:
    print(str(count) + '회차 i의 값: ')
    print(i)
    count = count + 1

# for문에서 2가지 이상의 값을 제어하는 방법
a = [(1, 1), (2, 2), (3, 3)]

for (b, c) in a:
    print(b + c)

# 2
# 4
# 6
````

## Range() 함수

#### 숫자 리스트를 자동으로 만들어주는 함수

- range(번호)
- range(시작번호, 마지막번호, Step)
    - Step은 생략이 가능함

````python
# range(번호)
for i in range(10):     # 0~9까지의 숫자 리스트 생성
    print(i)
# 0
# 1
# ...
# 8
# 9

print()

# range(시작번호, 마지막번호, Step)
# Step을 생략하면 1로 취급

for i in range(-10, 11):
    print(i)

# -10
# -8
# ...
# 8
# 10
````

|형태|설명|
|---|---|
|range(n)|0 부터 n-1까지|
|range(start,stop)|start부터 stop-1까지|
|range(start, stop, step)|step만큼 증가하면서 반복|

## while 반복문

#### 주어진 조건이 True(참)인 동안 코드를 반복적으로 실행하는 제어문

- 무한루프 위험이 있음
- 조건에 반드시 `변화`해야함

````python
num = 1

while num <= 5:
    print(num)
    num += 1        #num = num + 1
    # 첫번째 num = 1, 1 <= 5 문장이 참. print(num) 실행, num(1) + 1
    # 두번째 num = 2, 2 <= 5 문장이 참. print(num) 실행, num(2) + 1
    # 세번째 num = 3, ....
    # 다섯 번째 num = 5, 5 <= 5 문장이 참. print(num) 실행. num(5) + 1
    # 여섯 번째 num = 6, 6 <= 5 문장이 거짓. -> while문을 실행하지 않음.

print("while문이 종료된 후 num은 %d입니다." % num)
````

## Break문

반복문을 조건과 관계없이 강제로 탈출하는 명령어

- 반복 중단

````python
for i in range(10):
	if i == 5:
		break
	print(i)

"""
 0 ~ 9까지 반복
 i = 0
 i의 값은 5와 같지 않음(False) 반복 진행
 i = 2
 i의 값은 5와 같지 않음(False) 반복 진행
 ...
 i의 값은 5와 같음(True) break 실행 반복문 탈출
 0
 1
 2
 3
 4
"""
````

````python
num = 1

while True:
    print(num)
    num += 1

    if num > 5:
        break
"""
 첫번째 num = 1, True 문장이 참. print(num) 실행, num(1) + 1 -> if num(2) > 5: 거짓 -> pass
 두번째 num = 2, True 문장이 참. print(num) 실행, num(2) + 1 -> if num(3) > 5: 거짓 -> pass
 세번째 num = 3, ....
 다섯 번째 num = 5, True 문장이 참. print(num) 실행. num(5) + 1 -> if num(6) > 5: 참 -> if문에 있는 break 실행.
"""
````

## Continue 명령어

반복문을 continue 아래의 문장을 무시하고 다음 루프로 실행시키는 코드

- 현재 반복만 건너뛰고 다음 반복 진행

````python
for i in range(5):
    if i == 2:
        continue
    print(i)

"""
0 ~ 4까지 반복
 i = 0
 i의 값은 2와 같지 않음(False) 반복 진행
 i = 1
 i의 값은 2와 같지 않음(False) 반복 진행
 i = 2
 i의 값은 2와 같음(True) continue 실행 해당 반복문 생략 후 다음 반복 진행
 ...
 i = 4
 i의 값은 2와 같지 않음(False) 반복 진행
0
1
3
4
i == 2일때만 print를 실행하지 않음
"""
````

````python
num = 0

while num <= 10:

  num += 1      # num = num + 1     # num = 1

  if num < 5:
    continue        # num이 5미만일때 continue문으로 인해 print문이 실행되지 않는다.

  print(num)

"""
5
6
7
8
9
10
11
"""
````
