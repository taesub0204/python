# K-평균 군집화

보내주신 코드를 그대로 정리했습니다. 이미지 표시 및 군집화에는 컬러 RGB 정보를 사용합니다.

```python
import os
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
folder = './my'
files = [f for f in os.listdir(folder) if f.lower().endswith('.jpg')]

# print('이미지 개수:', len(files))

# img = Image.open(os.path.join(folder, files[0]))

# print('이미지 크기:', img.size)
```

```python
images = []

for file in files:
    path = os.path.join(folder, file)
    img = Image.open(path).convert('RGB')
    images.append(np.array(img))

images = np.array(images)
print(images.shape)
images_2d = images.reshape(-1, 100 * 100 * 3)
```

```python
def draw_fruits(arr, ratio=1): # arr = 112, n = 112, rows = 6, cols = 20
    n = len(arr)
    # n은 샘플 개수입니다.
    # 한 줄에 20개씩 이미지를 그립니다. 샘플 개수를 20으로 나누어 전체 행 개수를 계산합니다.
    rows = int(np.ceil(n / 20))
    # 행이 1개이면 열 개수는 샘플 개수입니다. 그렇지 않으면 20개입니다.
    cols = n if rows < 2 else 20
    fig, axs = plt.subplots(rows, cols,
                            figsize=(cols * ratio, rows * ratio), squeeze=False)
    for i in range(rows):
        for j in range(cols):
            if i * 20 + j < n:    # n개까지만 그립니다.
                axs[i, j].imshow(arr[i * 20 + j])
            axs[i, j].axis('off')
    plt.show()
```

```python
from sklearn.cluster import KMeans

# inertia = []

# for k in range(2, 10):
#     km = KMeans(n_clusters=k, random_state=42)
#     km.fit(images_2d)
#     inertia.append(km.inertia_)

# plt.plot(range(2, 10), inertia)
# plt.show()
```

```python
km = KMeans(n_clusters=7, random_state=42)
km.fit(images_2d)
print(km.labels_)
print(np.unique(km.labels_, return_counts=True))
```

```python
draw_fruits(images[km.labels_ == 5], ratio=2)
```
