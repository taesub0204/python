9. 장비별로 정상 범위를 벗어난 측정값이 몇 번 발생했는지 조회하시오.

```python
import pymysql
import pandas as pd
# 8. 각 장비가 몇 번의 공정을 수행했는지 조회하시오.

# # 1. DB 연걸



conn = pymysql.connect(
    host='localhost', 
    user='taesub', 
    password='0204',
    database='processsensordb',
    charset='utf8mb4' # 문자열을 UTF-8로 인코딩하여 저장하기 위해 설정
    )


query ="""
SELECT s.equipment_id, count(*) as abnormal
FROM sensormeasurement sm 
JOIN sensor s on sm.sensor_id = s.sensor_id
where sm.measured_value < s.normal_min
or sm.measured_value > s.normal_max
group by 
s.equipment_id
;
"""

df = pd.read_sql(query, conn)

print(df)
```

```python
import pymysql
import pandas as pd
# 파라미터 바인딩 연습문제 
# 연습문제 2.사용자로부터 equipment_id를 입력 받아, 해당 장비에서 수행된 공정 실행 이력을 조회하는 프로그램을 작성하시오

# # 1. DB 연걸



conn = pymysql.connect(
    host='localhost', 
    user='taesub', 
    password='0204',
    database='processsensordb',
    charset='utf8mb4' # 문자열을 UTF-8로 인코딩하여 저장하기 위해 설정
    )



equipment_id = input("조회할 equipment_id를 입력하세요: ")



query ="""


select run_id, e.equipment_id, e.model_name, p.step_name, r.start_time, r.end_time, r.run_status
from runhistory as r
join equipment as e
on r.equipment_id = e.equipment_id
join processstep as p 
on r.step_id = p.step_id
where e.equipment_id = %s
order by r.start_time
;
"""

df = pd.read_sql(query, conn, params=(equipment_id,))

print(df)
```
