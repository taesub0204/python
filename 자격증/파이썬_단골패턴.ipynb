# 파이썬 단골

```python
# 2차원 리스트 생성과 접근


matrix = [[1,2,3],
          [4,5,6],
          [7,8,9]
          
          ]
print(matrix[0])
print(matrix[1][2])


for row in matrix:
    for col in row:
        print(col, end='')
    print()
```

```python
def sum_2d(arr, i = 0, j =0):
    if i == len(arr):
        return 0
    if j == len(arr[i]):
        return sum_2d(arr, i+1, 0)
    return arr[i][j] + sum_2d(arr, i, j+1)
matrix = [[1,2,3],[4,5],[6,7,8]]
print(sum_2d(matrix))



```
